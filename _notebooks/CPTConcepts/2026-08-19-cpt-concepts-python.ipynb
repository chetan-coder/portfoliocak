## AlgorithmSequencing, selection **and** iteration together in one procedure. This is where the algorithm points are.

```python
# CODE_RUNNER: Average the part prices, then judge the build | height: 620pxdef calculate_average(numbers):    total = 0    count = 0    for num in numbers:              # ITERATION        total = total + num        count = count + 1    if count > 0:                    # SELECTION - no divide by zero        return total / count    else:        return 0# SEQUENCE - data, then process, then reportparts  = ["CPU", "GPU", "RAM", "Motherboard", "SSD", "PSU", "Case", "Cooler"]prices = [349, 599, 109, 179, 129, 119, 89, 99]average = calculate_average(prices)print("Average part price: $" + str(round(average, 2)))if average >= 250:    print("Build tier: Enthusiast")else:    print("Build tier: Mainstream or lower")print("")print("Parts pulling the average up:")for i in range(len(parts)):          # ITERATION over both lists at once    if prices[i] > average:          # SELECTION        print("  " + parts[i] + " at $" + str(prices[i]))
```
