## Iteration**CPT Requirement:** Your algorithm must include iteration (loops). College Board pseudocode offers `FOR EACH`, `REPEAT TIMES`, and `REPEAT UNTIL`; Python offers `for item in list:` and `while condition:`. Iteration processes multiple data items without writing duplicate code.**Python detail:** `REPEAT UNTIL (x = "no")` becomes `while x != "no":`—the condition **flips**. Pseudocode says "keep going until this is true," Python says "keep going while this is false." Getting that backwards produces an infinite loop, which on the Code Runner just times out.The pattern below is a **sentinel loop**: it runs an unknown number of times and stops on a magic value ("no"). For CPT, iteration shows your program can handle collections of data—critical for demonstrating algorithm complexity.

````python
# CODE_RUNNER: Keep adding parts to the cart until the builder says "no" | height: 620px# Canned keyboard. Each ask() pops the next answer off the front of this list.answers = ["GPU", 599, "yes", "CPU", 349, "yes", "PSU", 119, "no"]def ask(question):    answer = answers.pop(0)    print(question + " " + str(answer))    return answerprint("=== PC Build Part Entry ===")cart = []part_count = 0total_cost = 0keep_going = "yes"# REPEAT UNTIL (keep_going = "no")  ->  while keep_going != "no":while keep_going != "no":    part = ask("Add a part:")    price = int(ask("Price of the " + part + ":"))    cart.append(part + " ($" + str(price) + ")")   # APPEND    part_count = part_count + 1    total_cost = total_cost + price    keep_going = ask("Add another part? (no to stop)")print("")print("Cart: " + str(cart))print("Parts added: " + str(part_count))print("Total cost: $" + str(total_cost))
````

## Algorithm**CPT Requirement:** Your algorithm must integrate sequencing, selection, **and** iteration to solve a meaningful problem. College Board defines an algorithm as a precise step-by-step procedure that accomplishes a task. On the exam, you'll analyze complete algorithms for correctness and efficiency.For CPT, your algorithm is the core of your program—it must demonstrate complexity by combining multiple concepts to process data and produce results. **This is where you earn algorithm implementation points**, so the example below deliberately uses all three: a `for` loop (iteration), an `if/else` (selection), and ordered calculation steps (sequencing), all inside a procedure that takes a parameter and returns a value.

````python
# CODE_RUNNER: Average part price, then judge the build tier | height: 660pxdef calculate_average(numbers):    """Iteration + selection + return, all in one procedure."""    total = 0    count = 0    # ITERATION - walk the whole list    for num in numbers:        total = total + num        count = count + 1    # SELECTION - guard against dividing by zero on an empty list    if count > 0:        return total / count    else:        return 0# SEQUENCE - build the data, then process it, then report on itparts  = ["CPU", "GPU", "RAM", "Motherboard", "SSD", "PSU", "Case", "Cooler"]prices = [349, 599, 109, 179, 129, 119, 89, 99]average = calculate_average(prices)print("Parts: " + str(parts))print("Average part price: $" + str(round(average, 2)))# SELECTION on the resultif average >= 250:    print("Build tier: Enthusiast")else:    print("Build tier: Mainstream or lower")# ITERATION again, this time over both parallel lists at onceprint("")print("Parts pulling the average up:")for i in range(len(parts)):    if prices[i] > average:        print("  " + parts[i] + " at $" + str(prices[i]))
````

## List Operations**Exam Topic:** College Board pseudocode includes essential list operations: `APPEND(list, value)` adds to the end, `INSERT(list, index, value)` adds at a position, `REMOVE(list, index)` deletes by position, and `LENGTH(list)` returns the count.**Python equivalents:** `list.append(value)`, `list.insert(index, value)`, `list.pop(index)`, and `len(list)`. Remember the index shift—pseudocode's `INSERT(tasks, 2, x)` puts `x` in the second slot, but Python's `insert(2, x)` puts it in the **third**. On exam questions, you'll trace how these operations modify lists and affect indices. For CPT, manipulating lists demonstrates data management skills and algorithm complexity.

## Boolean Logic**Exam Concept:** College Board pseudocode uses `AND`, `OR`, and `NOT` to combine conditions; Python spells them `and`, `or`, and `not`. A single `=` is assignment, and Python will refuse to run `if price = 599:` at all—which is genuinely helpful, because in some languages that typo compiles and silently ruins your program.For CPT, Boolean logic enables sophisticated decision-making.
