## Output
**CPT Requirement:** Your program must produce output visible to the user. College Board pseudocode uses `DISPLAY()` to show results—strings, numbers, or variable values. On the exam, you'll trace DISPLAY statements to predict program output. For CPT, demonstrating clear output shows your program's purpose and functionality.

````python
# CODE_RUNNER: What each PC part does and what it costs | height: 560px

# Each part is stored under its own key
parts = {
    "CPU":  {"model": "Ryzen 7 7800X3D", "job": "Runs the OS and game logic",             "price": 349},
    "GPU":  {"model": "RTX 4070 Super",  "job": "Draws every pixel; also AI work",        "price": 599},
    "RAM":  {"model": "32GB DDR5-6000",  "job": "Short-term memory, wiped on power off",  "price": 109},
    "MOBO": {"model": "B650 ATX",        "job": "Connects it all; socket must match CPU", "price": 179},
    "SSD":  {"model": "2TB NVMe Gen4",   "job": "Long-term storage, survives power off",  "price": 129},
    "PSU":  {"model": "850W 80+ Gold",   "job": "Wall AC to clean DC; needs headroom",    "price": 119},
    "CASE": {"model": "Mid Tower ATX",   "job": "Holds the parts, controls airflow",       "price": 89},
    "COOL": {"model": "240mm AIO",       "job": "Moves heat off the CPU",                  "price": 99},
}

total = 0

# .items() hands you the key and the value together
for label, part in parts.items():
    model = part["model"]
    job   = part["job"]
    price = part["price"]

    # str() is required - you cannot add an int to a string in Python
    print(label.ljust(6) + model.ljust(18) + ("$" + str(price)).rjust(6) + "   " + job)
    total = total + price

print("")
print("TOTAL BUILD COST: $" + str(total))
````

## Input
**CPT Requirement:** Your program must get input from the user. College Board pseudocode uses `INPUT()` to collect data—this makes programs interactive and personalized. On exam questions, you'll identify where input affects program behavior. For CPT scoring, input demonstrates your program responds to different user data, a key component of functionality.

````python
# CODE_RUNNER: Check one part against your budget | height: 480px

# ===== INPUT - change these, then press Run =====
part_name = "GPU"      # try: CPU, GPU, RAM, PSU
budget    = 1500       # try: 400 or 2500
# ================================================

# Part name -> price
prices = {"CPU": 349, "GPU": 599, "RAM": 109, "PSU": 119}

print("Shopping for a " + part_name + " with a $" + str(budget) + " budget.")

if part_name in prices:
    price = prices[part_name]
    print("The " + part_name + " costs $" + str(price) + ".")
    print("Left over: $" + str(budget - price))
else:
    print("No part named " + part_name + " in this build.")
````

## Dictionary
**CPT Requirement:** Your program must use a dictionary (or other collection type) to manage multiple data values. A dictionary stores each value under a key, so a part is found by name. **Critical exam detail:** College Board pseudocode lists start at index 1, not 0. Accessing index 0 or going beyond LENGTH terminates the program. On exam questions, you'll trace collection operations and identify lookup errors. For CPT, using dictionaries to store and process data earns points for algorithm implementation and managing complexity.

````python
# CODE_RUNNER: Total the build from a dictionary of prices | height: 460px

# Part name -> price
build = {
    "CPU": 349,
    "GPU": 599,
    "RAM": 109,
    "Motherboard": 179,
    "SSD": 129,
    "PSU": 119,
    "Case": 89,
    "Cooler": 99,
}

total = 0

for part in build:                 # looping a dictionary walks its keys
    total = total + build[part]

print("Parts in build: " + str(len(build)))                # LENGTH(list)
print("Look up by name: CPU costs $" + str(build["CPU"]))
print("All the keys: " + ", ".join(build.keys()))
print("Priciest single part: $" + str(max(build.values())))
print("Total build cost: $" + str(total))

# build["Fan"] would raise a KeyError - test with "in", or use .get()
print("Is Fan in the build? " + str("Fan" in build))
print("build.get('Fan') is " + str(build.get("Fan")))
````

## Procedure
**CPT Requirement:** You must create at least one student-developed procedure with parameters. College Board pseudocode uses `PROCEDURE name(parameters)` with optional `RETURN(value)`. This demonstrates abstraction—hiding complexity and enabling code reuse. On the exam, you'll trace procedure calls and identify parameter passing. For CPT, your procedure must be called at least once and contribute to the program's functionality to earn full credit.

````python
# CODE_RUNNER: Sort each GPU into a price tier | height: 500px

def get_tier(price):
    """One parameter in, one value out."""
    if price >= 900:
        return "Enthusiast"
    else:
        if price >= 450:
            return "High-End"
        else:
            return "Budget"


# Model name -> price
gpus = {
    "RTX 4090": 1599,
    "RTX 4070 Super": 599,
    "RTX 4060": 299,
    "RX 7600": 249,
}

# One procedure, called four times - that is the point of abstraction
for gpu, price in gpus.items():
    print(gpu.ljust(16) + ("$" + str(price)).rjust(6) + "  ->  " + get_tier(price))
````

## Sequence
**Exam Concept:** Sequencing means statements execute in order, line by line. Every program uses sequence—it's the default flow. On exam questions, you'll trace execution order to predict variable values. For CPT, showing logical step-by-step calculations demonstrates you understand how programs process data systematically.


````python
# CODE_RUNNER: Checkout total, one step at a time | height: 500px

# ===== INPUT - change these, then press Run =====
rates = {"tax": 0.0775, "shipping": 24.99}           # tax is 7.75%
# ================================================

cart = {"CPU": 349, "GPU": 599, "PSU": 119}          # 1. what you are buying

subtotal = cart["CPU"] + cart["GPU"] + cart["PSU"]   # 2. needs the three above
tax = subtotal * rates["tax"]                        # 3. needs subtotal
final_cost = subtotal + tax + rates["shipping"]      # 4. needs both

print("CPU ........ $" + str(cart["CPU"]))
print("GPU ........ $" + str(cart["GPU"]))
print("PSU ........ $" + str(cart["PSU"]))
print("Subtotal ... $" + str(subtotal))
print("Tax ........ $" + str(round(tax, 2)))
print("Shipping ... $" + str(rates["shipping"]))
print("TOTAL ...... $" + str(round(final_cost, 2)))
````

## Selection
**CPT Requirement:** Your algorithm must include selection using `IF` or `IF/ELSE` statements. College Board pseudocode uses curly braces `{ }` to mark code blocks. Selection makes decisions based on Boolean conditions, allowing different outcomes. On the exam, you'll trace which branches execute based on condition values. For CPT, selection demonstrates your program handles different data appropriately—a key scoring criterion.


````python
# CODE_RUNNER: Pick a PSU from the GPU's power draw | height: 540px

# ===== INPUT - change this, then press Run =====
gpu_watts = 285        # try: 150, 285, 350, 500
# ===============================================

# Every PSU this shop stocks
psu_options = {
    "platinum1000": "1000W 80+ Platinum ($189)",
    "gold850":      "850W 80+ Gold ($119)",
    "gold750":      "750W 80+ Gold ($99)",
    "bronze550":    "550W 80+ Bronze ($59)",
}

# Nested - the exact shape of the College Board pseudocode
if gpu_watts >= 450:
    choice = "platinum1000"
else:
    if gpu_watts >= 300:
        choice = "gold850"
    else:
        if gpu_watts >= 200:
            choice = "gold750"
        else:
            choice = "bronze550"

print("GPU draws " + str(gpu_watts) + "W")
print("Recommended PSU: " + psu_options[choice])

# Same logic with elif - how Python is actually written
if gpu_watts >= 450:
    tier = "platinum1000"
elif gpu_watts >= 300:
    tier = "gold850"
elif gpu_watts >= 200:
    tier = "gold750"
else:
    tier = "bronze550"

notes = {
    "platinum1000": "top tier",
    "gold850":      "mid-high tier",
    "gold750":      "mid tier",
    "bronze550":    "entry tier",
}

print("Class: " + notes[tier])
````

## Iteration
**CPT Requirement:** Your algorithm must include iteration (loops) using `FOR EACH`, `REPEAT TIMES`, or `REPEAT UNTIL`. Iteration processes multiple data items without writing duplicate code. On the exam, you'll trace loop counters and predict how many times blocks execute. For CPT, iteration shows your program can handle collections of data—critical for demonstrating algorithm complexity and earning full points.

````python
# CODE_RUNNER: Add parts to a cart until the shopping list runs out | height: 560px

# ===== INPUT - edit this shopping list, then press Run =====
wanted = {"GPU": 1, "CPU": 1, "PSU": 1, "RAM": 2}    # part -> how many
# ===========================================================

catalog = {"CPU": 349, "GPU": 599, "RAM": 109, "Motherboard": 179, "SSD": 129, "PSU": 119}

cart = {}          # part -> what that line costs
total = 0

names = list(wanted.keys())
i = 0

# REPEAT UNTIL (i = LENGTH(wanted))  ->  while i != len(names)
while i != len(names):
    part = names[i]
    qty  = wanted[part]

    if part in catalog:
        line_cost = catalog[part] * qty
        cart[part] = line_cost                # APPEND
        total = total + line_cost
        print("Added " + str(qty) + " x " + part + " for $" + str(line_cost))
    else:
        print("Skipped " + part + " - not in the catalog")

    i = i + 1

print("")
print("Cart: " + ", ".join(cart.keys()))
print("Items: " + str(len(cart)))
print("Total: $" + str(total))
````

## Algorithm
**CPT Requirement:** Your algorithm must integrate sequencing, selection, and iteration to solve a meaningful problem. College Board defines an algorithm as a precise step-by-step procedure that accomplishes a task. On the exam, you'll analyze complete algorithms for correctness and efficiency. For CPT, your algorithm is the core of your program—it must demonstrate complexity by combining multiple concepts to process data and produce results. This is where you earn algorithm implementation points.

````python
# CODE_RUNNER: Average the part prices, then judge the build | height: 620px

def calculate_average(price_dict):
    total = 0
    count = 0

    for part in price_dict:          # ITERATION
        total = total + price_dict[part]
        count = count + 1

    if count > 0:                    # SELECTION - no divide by zero
        return total / count
    else:
        return 0


# SEQUENCE - data, then process, then report
build = {
    "CPU": 349,
    "GPU": 599,
    "RAM": 109,
    "Motherboard": 179,
    "SSD": 129,
    "PSU": 119,
    "Case": 89,
    "Cooler": 99,
}

average = calculate_average(build)
print("Average part price: $" + str(round(average, 2)))

if average >= 250:
    print("Build tier: Enthusiast")
else:
    print("Build tier: Mainstream or lower")

print("")
print("Parts pulling the average up:")
for part, price in build.items():    # ITERATION
    if price > average:              # SELECTION
        print("  " + part + " at $" + str(price))
````

## Dictionary Operations
**Exam Topic:** College Board pseudocode includes essential collection operations: `APPEND(list, value)` adds, `INSERT(list, index, value)` adds at a position, `REMOVE(list, index)` deletes, and `LENGTH(list)` returns count. In a dictionary these are keyed: assigning a new key adds, assigning an existing key updates, deleting a key removes, and counting the keys returns the length. On exam questions, you'll trace how these operations modify a collection and affect lookups. For CPT, manipulating dictionaries demonstrates data management skills and algorithm complexity.

````python
# CODE_RUNNER: Build a parts dictionary with add, update, and delete | height: 500px

parts = {"CPU": 349, "GPU": 599}
print("Start:            " + str(parts))

parts["PSU"] = 119               # a new key is added
print("add PSU:          " + str(parts))

parts["Motherboard"] = 179
print("add Motherboard:  " + str(parts))

parts["GPU"] = 549               # a key that already exists is updated
print("update GPU:       " + str(parts))

del parts["CPU"]                 # removed by key
print("del CPU:          " + str(parts))

print("number of keys:   " + str(len(parts)))            # LENGTH(list)
print("has a Cooler?     " + str("Cooler" in parts))

print("")
for part, price in parts.items():
    print("Installing: " + part + " for $" + str(price))
````

## Search Algorithm
**Exam Topic:** Linear search is a common algorithm pattern that appears frequently on the AP exam. It demonstrates iteration through a dictionary, selection to check conditions, and return values. On exam questions, you'll identify search algorithms, trace their execution, and determine what they return when items are found or not found. For CPT, implementing search shows algorithmic thinking and practical data processing.

````python
# CODE_RUNNER: Find a part and its price | height: 560px

# ===== INPUT - change this, then press Run =====
search_for = "RAM"     # try: CPU, GPU, RAM, PSU, or Fan
# ===============================================

def find_price(price_dict, target):
    """Linear search walks the keys one at a time."""
    for part in price_dict:
        if part == target:             # == compares, = assigns
            return price_dict[part]    # found - stop early

    return -1                          # sentinel meaning "not found"


build = {"CPU": 349, "GPU": 599, "RAM": 109, "PSU": 119}

price = find_price(build, search_for)

if price >= 0:
    print(search_for + " found by linear search")
    print("Price: $" + str(price))
else:
    print(search_for + " is not in this build")

# The same answer from a single lookup
if search_for in build:
    print("Direct lookup: $" + str(build[search_for]))
else:
    print("Direct lookup: no key named " + search_for)
````

## Boolean Logic
**Exam Concept:** College Board pseudocode uses `AND`, `OR`, and `NOT` operators to combine conditions. Boolean expressions evaluate to true or false. On the exam, you'll trace complex conditions and predict which code blocks execute. For CPT, boolean logic enables sophisticated decision-making—like role-based access control or validation checks—demonstrating program complexity.

````python
# CODE_RUNNER: Will these parts actually work together? | height: 620px

# ===== INPUT - change these, then press Run =====
build = {
    "cpu_socket":   "AM5",     # try "LGA1700" to break it
    "board_socket": "AM5",
    "psu_watts":    850,       # try 400
    "gpu_watts":    285,
    "gpu_length":   304,       # mm - try 400
    "case_space":   340,       # mm
}
# ================================================

# Each check holds a True or a False
checks = {
    "socket": build["cpu_socket"] == build["board_socket"],
    "power":  build["psu_watts"] >= build["gpu_watts"] + 200,   # 200W of headroom
    "fits":   build["gpu_length"] <= build["case_space"],
}

for name, result in checks.items():
    print(name.ljust(8) + " ok? " + str(result))
print("")

# AND - every condition must be True
if checks["socket"] and checks["power"] and checks["fits"]:
    print("COMPATIBLE - this build will boot.")
else:
    print("INCOMPATIBLE - fix it before ordering.")

# OR - one True is enough to warn
if not checks["power"] or not checks["fits"]:
    print("Warning: check the PSU or the GPU clearance.")

# NOT - flips a condition
if not checks["socket"]:
    print("The CPU will not physically drop into that board.")
````
