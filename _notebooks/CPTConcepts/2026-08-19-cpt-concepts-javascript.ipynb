## Sequence**Exam Concept:** Sequencing means statements execute in order, line by line. Every program uses sequence—it's the default flow. ## Iteration**CPT Requirement:** Your algorithm must include iteration (loops). Getting that backwards produces an infinite loop, which in a browser freezes the whole tab.The pattern below is a **sentinel loop**: it runs an unknown number of times and stops on a magic value (`"no"`). ## List Operations**Exam Topic:** College Board pseudocode includes essential list operations: `APPEND(list, value)` adds to the end, `INSERT(list, index, value)` adds at a position, `REMOVE(list, index)` deletes by position, and `LENGTH(list)` returns the count.**JavaScript equivalents:** `array.push(value)`, `array.splice(index, 0, value)`, `array.splice(index, 1)`, and `array.length`. Remember the index shift—pseudocode's `INSERT(tasks, 2, x)` puts `x` in the second slot, but JavaScript's `splice(2, 0, x)` puts it in the **third**. On exam questions, you'll trace how these operations modify lists and affect indices. For CPT, manipulating lists demonstrates data management skills and algorithm complexity.

````python
%%js// CODE_RUNNER: Linear search for a part in the build | height: 580px// Walk the array one element at a time; return the index or -1function findItem(items, target) {  let index = 0;                   // JavaScript arrays start at 0, not 1  for (const item of items) {    if (item === target) {         // === compares, = assigns      return index;                // found it - stop early, don't finish the loop    }    index = index + 1;  }  return -1;                       // sentinel value meaning "not found"}const parts = ["CPU", "GPU", "RAM", "PSU"];const searchFor = "RAM";// Real version:  const searchFor = document.getElementById("searchInput").value;console.log("Searching for: " + searchFor);const position = findItem(parts, searchFor);if (position >= 0) {  console.log("Found at index " + position + " (that is part #" + (position + 1) + ")");} else {  console.log("Not in this build");}// JavaScript has a built-in that does the same job in one lineconsole.log("Built-in check with indexOf(): " + parts.indexOf(searchFor));
````

## Boolean Logic**Exam Concept:** College Board pseudocode uses `AND`, `OR`, and `NOT` to combine conditions; JavaScript spells them `&&`, `||`, and `!`. Boolean expressions evaluate to `true` or `false`. On the exam, you'll trace complex conditions and predict which code blocks execute.**JavaScript detail:** comparison is `===` (**triple** equals). Double `==` also exists but converts types behind your back—`0 == "0"` is `true`, which causes bugs that take hours to find. Always use `===`. And a single `=` inside an `if` is an assignment that JavaScript will happily run, silently overwriting your variable.For CPT, Boolean logic enables sophisticated decision-making. In a PC build it answers the only question that matters before you spend $1,672: **will these parts actually work together?**

````python
%%js// CODE_RUNNER: Look up a price, then test it with the AND, OR, NOT operators | height: 700pxfunction findPartPrice(parts, prices, targetName) {  let index = 0;  for (const part of parts) {    if (part === targetName) {      return prices[index];    }    index = index + 1;  }  return -1;}const parts  = ["CPU", "GPU", "RAM", "PSU"];const prices = [349, 599, 109, 119];const searchName = "GPU";const budget = 500;console.log("Looking up: " + searchName + " against a $" + budget + " budget");const price = findPartPrice(parts, prices, searchName);console.log("Price found: $" + price);console.log("");// AND (&&) - every condition must be trueif (price > 0 && price <= budget) {  console.log("&&  -> in the list AND within budget");} else {  console.log("&&  -> failed: it is either missing or too expensive");}// OR (||) - one true is enoughif (price > budget || price === -1) {  console.log("||  -> flagged: over budget, or not in the build at all");}// NOT (!) - flips a conditionif (!(price === -1)) {  console.log("!   -> confirmed present in this build");}console.log("");console.log("--- Full compatibility check ---");const cpuSocket     = "AM5";const boardSocket   = "AM5";const psuWatts      = 850;const gpuWatts      = 285;const gpuLength     = 304;    // millimetresconst caseClearance = 340;    // millimetresconst socketOk = cpuSocket === boardSocket;const powerOk  = psuWatts >= gpuWatts + 200;const fitsOk   = gpuLength <= caseClearance;console.log("Socket match (AM5 = AM5): " + socketOk);console.log("Power headroom >= 200W:   " + powerOk);console.log("GPU fits the case:        " + fitsOk);if (socketOk && powerOk && fitsOk) {  console.log("RESULT: compatible - this build will boot.");} else {  console.log("RESULT: incompatible - fix the parts list before ordering.");}
````
