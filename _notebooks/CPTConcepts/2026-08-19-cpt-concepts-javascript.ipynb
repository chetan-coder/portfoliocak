## IterationYour algorithm needs a loop. `REPEAT UNTIL (x = "no")` becomes `while (x !== "no")` — the condition flips.

```python
%%js// CODE_RUNNER: Add parts to a cart until the list runs out | height: 560px// ===== INPUT - edit this shopping list, then press Run =====const wanted = ["GPU", "CPU", "PSU", "RAM"];// ===========================================================const catalog = ["CPU", "GPU", "RAM", "Motherboard", "SSD", "PSU"];const costs   = [349, 599, 109, 179, 129, 119];let cart = [];let total = 0;let i = 0;// REPEAT UNTIL (i = LENGTH(wanted))  ->  while (i !== wanted.length)while (i !== wanted.length) {  const part = wanted[i];  if (catalog.includes(part)) {    const price = costs[catalog.indexOf(part)];    cart.push(part);                       // APPEND    total = total + price;    console.log("Added " + part + " for $" + price);  } else {    console.log("Skipped " + part + " - not in the catalog");  }  i = i + 1;}console.log("");console.log("Cart: " + cart.join(", "));console.log("Items: " + cart.length);console.log("Total: $" + total);
```

## AlgorithmSequencing, selection **and** iteration together in one procedure. This is where the algorithm points are.

```python
%%js// CODE_RUNNER: Average the part prices, then judge the build | height: 620pxfunction calculateAverage(numbers) {  let total = 0;  let count = 0;  for (const num of numbers) {       // ITERATION    total = total + num;    count = count + 1;  }  if (count > 0) {                   // SELECTION - no divide by zero    return total / count;  } else {    return 0;  }}// SEQUENCE - data, then process, then reportconst parts  = ["CPU", "GPU", "RAM", "Motherboard", "SSD", "PSU", "Case", "Cooler"];const prices = [349, 599, 109, 179, 129, 119, 89, 99];const average = calculateAverage(prices);console.log("Average part price: $" + average.toFixed(2));if (average >= 250) {  console.log("Build tier: Enthusiast");} else {  console.log("Build tier: Mainstream or lower");}console.log("");console.log("Parts pulling the average up:");for (let i = 0; i < parts.length; i++) {   // ITERATION over both arrays  if (prices[i] > average) {               // SELECTION    console.log("  " + parts[i] + " at $" + prices[i]);  }}
```

## Search AlgorithmLinear search: walk the list, compare, return early. Return `-1` when nothing matches.

```python
%%js// CODE_RUNNER: Find a part and its price | height: 560px// ===== INPUT - change this, then press Run =====const searchFor = "RAM";    // try: CPU, GPU, RAM, PSU, or Fan// ===============================================function findIndex(items, target) {  let index = 0;                 // arrays start at 0, not 1  for (const item of items) {    if (item === target) {       // === compares, = assigns      return index;              // found - stop early    }    index = index + 1;  }  return -1;                     // sentinel meaning "not found"}const parts  = ["CPU", "GPU", "RAM", "PSU"];const prices = [349, 599, 109, 119];const position = findIndex(parts, searchFor);if (position >= 0) {  const price = prices[position];   // index from one array reads the other  console.log(searchFor + " found at index " + position);  console.log("Price: $" + price);} else {  console.log(searchFor + " is not in this build");}
```

## Boolean Logic`AND` / `OR` / `NOT` become `&&` / `||` / `!`. A PC build is a compatibility problem, so it is all booleans.

```python
%%js// CODE_RUNNER: Will these parts actually work together? | height: 620px// ===== INPUT - change these, then press Run =====const cpuSocket   = "AM5";   // try "LGA1700" to break itconst boardSocket = "AM5";const psuWatts    = 850;     // try 400const gpuWatts    = 285;const gpuLength   = 304;     // mm - try 400const caseSpace   = 340;     // mm// ================================================const socketOk = cpuSocket === boardSocket;const powerOk  = psuWatts >= gpuWatts + 200;   // 200W of headroomconst fitsOk   = gpuLength <= caseSpace;console.log("Socket match: " + socketOk);console.log("Power ok:     " + powerOk);console.log("GPU fits:     " + fitsOk);console.log("");// && - every condition must be trueif (socketOk && powerOk && fitsOk) {  console.log("COMPATIBLE - this build will boot.");} else {  console.log("INCOMPATIBLE - fix it before ordering.");}// || - one true is enough to warnif (!powerOk || !fitsOk) {  console.log("Warning: check the PSU or the GPU clearance.");}// ! - flips a conditionif (!socketOk) {  console.log("The CPU will not physically drop into that board.");}
```
